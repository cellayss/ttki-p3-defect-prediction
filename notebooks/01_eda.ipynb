Memeriksa setiap file proyek dari Kaggle sebelum eksperimen. Hasilnya disalin ke `data/README.md` dan bagian Methods.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
from src import preprocessing as pp, evaluation as ev
import pandas as pd
```

```python
files = pp.list_project_files('../data/raw')
[f.name for f in files]
```

```python
rows = []
for f in files:
    df = pp.load_file(f)
    label = pp.find_column(df, pp.LABEL_CANDIDATES)
    loc = pp.find_column(df, pp.LOC_CANDIDATES)
    print(f.name, '| label =', label, '| LOC =', loc, '| kolom:', list(df.columns))
    if label and loc:
        rows.append({'project': f.stem, **pp.audit(df, label, loc)})
audit = pd.DataFrame(rows)
audit
```

**Cocokkan jumlah baris** dengan Tabel I Shepperd et al. (2013) untuk menentukan varian (MDP asli / PROMISE / D' / D'').

Contoh dari Shepperd: CM1 = 505 baris (MDP) vs 498 (PROMISE); KC1 = 2.107 baris (MDP), JM1 = 10.878 baris (MDP).

```python
cleaning = []
for f in files:
    df = pp.load_file(f)
    label = pp.find_column(df, pp.LABEL_CANDIDATES)
    if label:
        _, log = pp.clean(df, label)
        cleaning.append({'project': f.stem, **log})
pd.DataFrame(cleaning)
```
